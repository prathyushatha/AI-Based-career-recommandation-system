```python
import pandas as pd
from collections import Counter
import re

df = pd.read_csv(
    "../data/processed_career_dataset.csv"
)

career_list = list(
    df["Career"].unique()
)

career_skill_map = {}

for career in career_list:

    career_df = df[
        df["Career"] == career
    ]

    skills = []

    for row in career_df["Skills"]:

        skill_list = re.split(
            r",|\s{2,}",
            str(row)
        )

        skill_list = [

            s.strip().lower()

            for s in skill_list

            if s.strip() != ""

        ]

        skills.extend(skill_list)

    career_skill_map[career] = [

        skill

        for skill, count in Counter(
            skills
        ).most_common(10)

    ]

def detect_career(question):

    question = question.lower()

    for career in career_list:

        if career.lower() in question:

            return career

    return None

def chatbot(question):

    career = detect_career(question)

    if career is None:

        return "Please mention a valid career."

    question = question.lower()

    if "skill" in question:

        return (
            "Required Skills:\n"
            + ", ".join(
                career_skill_map[career]
            )
        )

    elif "roadmap" in question:

        return (
            f"Roadmap for {career}:\n"
            "1. Learn Fundamentals\n"
            "2. Build Projects\n"
            "3. Practice DSA\n"
            "4. Create Portfolio\n"
            "5. Apply for Internships"
        )

    elif "salary" in question:

        return (
            f"{career} salary depends on company, location and experience."
        )

    elif "future" in question:

        return (
            f"{career} has excellent future demand."
        )

    else:

        return (
            "I can answer questions about skills, roadmap, salary and future."
        )

print(
    chatbot(
        "What skills are needed for Data Scientist?"
    )
)

print()

print(
    chatbot(
        "What is the roadmap for ML Engineer?"
    )
)
```
