```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\balaj\OneDrive\Documents\AI_Career_Recommendation_System\data\Combined_AI_Career_Recommendation_Dataset.csv"
)
df.isnull().sum()
text_cols = [
    'Gender',
    'Location',
    'Highest_Qualification',
    'Stream',
    'Academic_Level',
    'Soft_Skills',
    'Languages_Known',
    'Certifications',
    'Preferred_Work_Style',
    'Work_Type_Interest',
    'Past_Jobs_Internships',
    'Achievements',
    'Skills_Gained',
    'Willing_to_Relocate'
]

for col in text_cols:
    df[col] = df[col].fillna('Unknown')

df['Age'] = df['Age'].fillna(
    df['Age'].median()
)
df.isnull().sum()
def clean_text(text):

    text = str(text).lower()

    text = text.replace(",", " ")

    return text
text_columns = [
    'Skills',
    'Interest',
    'Soft_Skills',
    'Certifications',
    'Skills_Gained'
]

for col in text_columns:

    df[col] = df[col].apply(clean_text)

df["student_profile"] = (
    df["Skills"] + " " +
    df["Interest"] + " " +
    df["Soft_Skills"] + " " +
    df["Certifications"] + " " +
    df["Skills_Gained"]
)

df.to_csv(
    "../data/processed_career_dataset.csv",
    index=False
)
```
