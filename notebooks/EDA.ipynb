```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\balaj\OneDrive\Documents\AI_Career_Recommendation_System\data\Combined_AI_Career_Recommendation_Dataset.csv"
)

print(df.head())
print(df.shape)
print(df.isnull().sum())
df.info()
import matplotlib.pyplot as plt

missing = df.isnull().sum()

missing = missing[missing > 0]

plt.figure(figsize=(10,6))

missing.sort_values().plot(kind='barh')

plt.title("Missing Values")

plt.show()
plt.figure(figsize=(12,8))

df['Career'].value_counts().plot(
    kind='bar'
)

plt.title("Career Distribution")

plt.show()
plt.figure(figsize=(10,6))

df['Interest'].value_counts().plot(
    kind='bar'
)

plt.title("Interest Distribution")

plt.show()

plt.figure(figsize=(10,6))

df['Highest_Qualification']\
.value_counts()\
.plot(kind='bar')

plt.title(
    "Highest Qualification Distribution"
)

plt.show()
plt.figure(figsize=(10,6))

df['Stream']\
.value_counts()\
.plot(kind='bar')

plt.title("Academic Stream Distribution")

plt.show()
plt.figure(figsize=(8,5))

df['Preferred_Work_Style']\
.value_counts()\
.plot(kind='pie',
      autopct='%1.1f%%')

plt.title(
    "Work Style Preference"
)

plt.show()
plt.figure(figsize=(8,5))

df['Willing_to_Relocate']\
.value_counts()\
.plot(kind='pie',
      autopct='%1.1f%%')

plt.title(
    "Relocation Preference"
)

plt.show()
plt.figure(figsize=(10,6))

df['CGPA'].hist(
    bins=20
)

plt.title(
    "CGPA Distribution"
)

plt.xlabel("CGPA")

plt.ylabel("Students")

plt.show()
plt.figure(figsize=(15,6))

import seaborn as sns

sns.boxplot(
    x='Career',
    y='CGPA',
    data=df
)

plt.xticks(rotation=90)

plt.show()
from collections import Counter

all_skills = []

for row in df['Skills']:

    skills = str(row).split(',')

    all_skills.extend(
        [skill.strip()
         for skill in skills]
    )

skill_counts = Counter(all_skills)

top_skills = dict(
    skill_counts.most_common(15)
)

plt.figure(figsize=(12,6))

plt.bar(
    top_skills.keys(),
    top_skills.values()
)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.title(
    "Top Skills"
)

plt.show()
```
