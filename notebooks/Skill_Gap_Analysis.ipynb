```python
import pandas as pd
from collections import Counter
import re

df = pd.read_csv(
    "../data/processed_career_dataset.csv"
)

career_skill_map = {}

for career in df["Career"].unique():

    career_df = df[
        df["Career"] == career
    ]

    all_skills = []

    for row in career_df["Skills"]:

        skills = re.split(
            r",|\s{2,}",
            str(row)
        )

        skills = [
            skill.strip().lower()
            for skill in skills
            if skill.strip() != ""
        ]

        all_skills.extend(skills)

    career_skill_map[career] = [

        skill

        for skill, count in Counter(
            all_skills
        ).most_common(10)

    ]

print(career_skill_map["Data Scientist"])


def skill_gap_analysis(
    user_skills,
    target_career
):

    required = career_skill_map[
        target_career
    ]

    missing = []

    for skill in required:

        if skill.lower() not in [

            s.lower()

            for s in user_skills

        ]:

            missing.append(skill)

    return missing


user_skills = [
    "python",
    "sql"
]

missing = skill_gap_analysis(
    user_skills,
    "Data Scientist"
)

print(missing)

learning_resources = {

"python":"Python → OOP → NumPy → Pandas",

"sql":"SQL → Joins → Aggregations",

"java":"Core Java → OOP → Collections",

"c++":"C++ → OOP → STL → DSA",

"javascript":"JavaScript → DOM → ES6",

"react":"React → Hooks → Projects",

"node.js":"Node.js → Express",

"html/css":"HTML → CSS → Responsive Design",

"machine learning":"Scikit-Learn → ML Projects",

"deep learning":"TensorFlow → PyTorch",

"data analysis":"Excel → SQL → Pandas"

}

def generate_learning_plan(
    missing_skills
):

    roadmap = {}

    for skill in missing_skills:

        roadmap[skill] = learning_resources.get(

            skill,

            f"Learn {skill}"

        )

    return roadmap

plan = generate_learning_plan(
    missing
)

for skill, path in plan.items():

    print()

    print(skill.upper())

    print(path)
```
