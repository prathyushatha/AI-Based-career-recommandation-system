```python
import pandas as pd
from collections import Counter
import re

df = pd.read_csv(
    "../data/processed_career_dataset.csv"
)

career_list = list(
    df["Career"].unique()
)

career_skill_map = {}

for career in career_list:

    career_df = df[
        df["Career"] == career
    ]

    skills = []

    for row in career_df["Skills"]:

        skill_list = re.split(
            r",|\s{2,}",
            str(row)
        )

        skill_list = [

            s.strip().lower()

            for s in skill_list

            if s.strip() != ""

        ]

        skills.extend(skill_list)

    career_skill_map[career] = [

        skill

        for skill, count in Counter(
            skills
        ).most_common(10)

    ]

def detect_career(question):

    question = question.lower()

    for career in career_list:

        if career.lower() in question:

            return career

    return None

def chatbot(question):

    career = detect_career(question)

    if career is None:

        return "Career not found."

    if "skill" in question.lower():

        return (

            f"Skills required for {career}:\n\n"

            + ", ".join(
                career_skill_map[career]
            )

        )

    return "Ask about skills."

print(
    chatbot(
        "What skills are required for Data Scientist?"
    )
)
```
